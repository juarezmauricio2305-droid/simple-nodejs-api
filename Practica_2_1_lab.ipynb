```python
# ==========================================
# PROYECTO 4: Agrupación de datos (.groupby)
# Equipo C - Data Science Showcase
# ==========================================

import pandas as pd
import numpy as np

# 1. Creación/Carga del Dataset Adaptado (Ventas e-commerce)
np.random.seed(42)
n_registros = 500

categorias = ['Electrónica', 'Ropa', 'Hogar', 'Juguetes']
regiones = ['Norte', 'Sur', 'Este', 'Oeste']

df_ventas = pd.DataFrame({
    'ID_Cliente': np.random.randint(1000, 2000, size=n_registros),
    'Categoria': np.random.choice(categorias, size=n_registros),
    'Region': np.random.choice(regiones, size=n_registros),
    'Monto_Venta': np.round(np.random.uniform(10.0, 500.0, size=n_registros), 2),
    'Cantidad': np.random.randint(1, 10, size=n_registros)
})

print("--- Vista previa del Dataset ---")
print(df_ventas.head())

# 2. Agrupación simple (.groupby)
print("\n--- 1. Venta promedio por Categoría ---")
promedio_categoria = df_ventas.groupby('Categoria')['Monto_Venta'].mean()
print(promedio_categoria)

# 3. Agrupación múltiple y agregaciones avanzadas (.agg)
print("\n--- 2. Resumen por Categoría y Región ---")
resumen_avanzado = df_ventas.groupby(['Categoria', 'Region']).agg(
    Ventas_Totales=('Monto_Venta', 'sum'),
    Promedio_Venta=('Monto_Venta', 'mean'),
    Total_Articulos=('Cantidad', 'sum'),
    Num_Transacciones=('Monto_Venta', 'count')
).reset_index()

print(resumen_avanzado.head(10))
```

```python
# ==========================================
# PROYECTO 5: Tablas Dinámicas (pivot_table)
# Equipo C - Data Science Showcase
# ==========================================

import pandas as pd
import numpy as np

# 1. Generación del Dataset Adaptado (Desempeño Anual por Región)
np.random.seed(10)
fechas = pd.date_range(start='2022-01-01', end='2025-12-31', freq='M')

datos = []
for fecha in fechas:
    for region in ['América Latina', 'Norteamérica', 'Europa', 'Asia']:
        datos.append({
            'Fecha': fecha,
            'Año': fecha.year,
            'Region': region,
            'Ingresos': np.random.randint(20000, 100000),
            'Gastos_Mkt': np.random.randint(5000, 25000)
        })

df_financiero = pd.DataFrame(datos)

# 2. Creación de Tabla Dinámica (Pivot Table) básica
print("--- Tabla Dinámica: Ingresos Totales por Región y Año ---")
pivot_ingresos = pd.pivot_table(
    df_financiero,
    values='Ingresos',
    index='Region',
    columns='Año',
    aggfunc='sum',
    margins=True,          # Incluye fila y columna de Totales
    margins_name='Total'
)
print(pivot_ingresos)

# 3. Tabla Dinámica con múltiples métricas agregadas
print("\n--- Tabla Dinámica: Promedio de Ingresos y Gastos ---")
pivot_multimetrica = pd.pivot_table(
    df_financiero,
    values=['Ingresos', 'Gastos_Mkt'],
    index=['Region'],
    columns=['Año'],
    aggfunc={'Ingresos': 'mean', 'Gastos_Mkt': 'sum'}
)
print(pivot_multimetrica)
```

```python
# ==========================================
# PROYECTO 6: Unión de DataFrames
# Equipo C - Data Science Showcase
# ==========================================

import pandas as pd

# 1. Creación de datasets relacionales de prueba
df_clientes = pd.DataFrame({
    'ID_Cliente': [101, 102, 103, 104, 105],
    'Nombre': ['Ana Gómez', 'Carlos López', 'María Silva', 'Jorge Hernández', 'Lucía Ruiz'],
    'Ciudad': ['CDMX', 'Guadalajara', 'Monterrey', 'CDMX', 'Puebla']
})

df_pedidos = pd.DataFrame({
    'ID_Pedido': [5001, 5002, 5003, 5004, 5005],
    'ID_Cliente': [101, 103, 101, 105, 106], # 106 es un cliente sin registro previo
    'Monto': [250.50, 120.00, 430.00, 89.90, 600.00],
    'Metodo_Pago': ['Tarjeta', 'Efectivo', 'Tarjeta', 'Transferencia', 'Tarjeta']
})

df_pedidos_sucursal_2 = pd.DataFrame({
    'ID_Pedido': [6001, 6002],
    'ID_Cliente': [102, 104],
    'Monto': [310.00, 150.00],
    'Metodo_Pago': ['Efectivo', 'Tarjeta']
})

# 2. Combinación vertical de registros con pd.concat()
print("--- 1. Concatenación de Pedidos de Diferentes Sucursales ---")
df_todos_los_pedidos = pd.concat([df_pedidos, df_pedidos_sucursal_2], ignore_index=True)
print(df_todos_los_pedidos)

# 3. Cruce de datos relacionales con pd.merge()
print("\n--- 2. Merge tipo Inner Join (Solo coincidencias) ---")
df_inner = pd.merge(df_clientes, df_todos_los_pedidos, on='ID_Cliente', how='inner')
print(df_inner[['ID_Cliente', 'Nombre', 'ID_Pedido', 'Monto']])

print("\n--- 3. Merge tipo Left Join (Todos los clientes con sus pedidos) ---")
df_left = pd.merge(df_clientes, df_todos_los_pedidos, on='ID_Cliente', how='left')
print(df_left[['ID_Cliente', 'Nombre', 'ID_Pedido', 'Monto']])
```
