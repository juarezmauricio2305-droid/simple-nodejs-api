# La estrategia Split-Apply-Combine

```python
# Let's categorize phones based on Price_in_thousands for demonstration
# split
high_end_filter = df_phones['Price_in_thousands'] >= 0.700
mid_range_filter = (df_phones['Price_in_thousands'] < 0.700) & (df_phones['Price_in_thousands'] >= 0.400)
low_end_filter = df_phones['Price_in_thousands'] < 0.400
```

```python
# apply
high_end_avg_sales = df_phones[high_end_filter]['Sales_in_thousands'].mean()
mid_range_avg_sales = df_phones[mid_range_filter]['Sales_in_thousands'].mean()
low_end_avg_sales = df_phones[low_end_filter]['Sales_in_thousands'].mean()
```

```python
# combine
pd.DataFrame({'Phone_Category':['High-End', 'Mid-Range', 'Low-End'],
              'Sales_in_thousands':[high_end_avg_sales, mid_range_avg_sales, low_end_avg_sales]}).set_index('Phone_Category')
```
